```python
import time
```

```python
def fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
```

```python
start=time.time()
```

```python
print (fib(33))
print(time.time()-start)
```

```python
start2=time.time()
```

```python
def fib(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fib(n-1, d) + fib(n-2, d)
        return d[n]

d = {0: 1, 1: 1}

print(fib(33, d))
print(time.time() - start2)
```

## The time difference is clear in case of memoization the time taken to calculate the fib series is far less , comapared to the authentic way
