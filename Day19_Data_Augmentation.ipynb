```python
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img=image.load_img('CAT.webp',target_size=(200,200))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(img)
```

```python
type(img)
```

```python
datagen=ImageDataGenerator(rotation_range=30,shear_range=0.2,zoom_range=0.2,horizontal_flip=True,width_shift_range=0.2,height_shift_range=0.2)
```

```python
img=image.img_to_array(img)
```

```python
type(img)
```

```python
print(img)
```

```python
img.shape
```

```python
input_img=img.reshape((1,200,200,3))
```

```python
input_img.shape
```

```python
i=0

for output in datagen.flow(input_img,batch_size=1,save_to_dir='mydir'):
    i+=1
    if i == 10:
        break
```
