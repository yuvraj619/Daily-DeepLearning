```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("DailyDelhiClimateTest.csv")
```

```python
df.head()
```

```python
df=df[['date','meantemp']]
```

```python
df.head()
```

```python
plt.figure(figsize=(20, 10))

plt.scatter(df.date, df.meantemp, c='red')

plt.xticks(rotation=90)

plt.xlabel('Date')
plt.ylabel('Mean Temperature')

plt.show()
```

```python
x1=df['meantemp'].ewm(alpha=0.2).mean()
```

```python
print(x1)
```

```python
df['EWM']=x1
```

```python
df.head()
```

```python
plt.figure(figsize=(20,10))
plt.scatter(df.date, df.meantemp, c='red')
plt.plot(df.date,df.EWM,c='green',label='alpha=0.1')

plt.xticks(rotation=90)

plt.xlabel('Date')
plt.ylabel('Mean Temperature')
plt.legend()

plt.show()
```
