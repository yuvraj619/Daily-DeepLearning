```python
import kagglehub

path = kagglehub.dataset_download("salader/dogsvscats")

print("Path to dataset files:", path)
```

```python
import os

print(os.listdir(path))
```

```python
print(os.listdir(os.path.join(path, "train")))
print(os.listdir(os.path.join(path, "test")))
```

```python
train_cats = os.listdir(os.path.join(path, "train", "cats"))
train_dogs = os.listdir(os.path.join(path, "train", "dogs"))

test_cats = os.listdir(os.path.join(path, "test", "cats"))
test_dogs = os.listdir(os.path.join(path, "test", "dogs"))

print("Training cats:", len(train_cats))
print("Training dogs:", len(train_dogs))
print("Testing cats:", len(test_cats))
print("Testing dogs:", len(test_dogs))
```

```python
import tensorflow as tf
from tensorflow import keras
import os

train_ds = keras.utils.image_dataset_from_directory(
    directory=os.path.join(path, "train"),
    labels="inferred",
    label_mode="int",
    batch_size=20000,
    image_size=(150, 150)
)

validation_ds = keras.utils.image_dataset_from_directory(
    directory=os.path.join(path, "test"),
    labels="inferred",
    label_mode="int",
    batch_size=32,
    image_size=(150, 150)
)
```

```python
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,Flatten
from keras.applications.vgg16 import VGG16
```

```python
conv_base=VGG16(weights='imagenet',include_top=False,input_shape=(150,150,3))
```

```python
conv_base.summary()
```

```python
model=Sequential()

model.add(conv_base)
model.add(Flatten())
model.add(Dense(256))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
conv_base.trainable=False
```

```python
model.summary()
```

```python
### normalization
def process(image,label):
    image = tf.cast(image/255. ,tf.float32)
    return image,label

train_ds = train_ds.map(process)
validation_ds = validation_ds.map(process)
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(train_ds,epochs=10,validation_data=validation_ds)
```
