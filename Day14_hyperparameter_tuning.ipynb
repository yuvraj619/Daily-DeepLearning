```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('diabetes.csv')
df.head()
```

```python
df.corr()['Outcome']
```

```python
X=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X=scaler.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test , y_train,y_test=train_test_split(X,y,random_state=1,test_size=0.2)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
```

```python
model_1=Sequential()
model_1.add(Input(shape=(8,)))
model_1.add(Dense(32,activation='relu'))
model_1.add(Dense(1,activation='sigmoid'))

model_1.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model_1.summary()
```

```python
model_1.fit(X_train,y_train,batch_size=32,epochs=100,validation_data=(X_test,y_test))
```

## TO CHECK FOR BEST OPTIMIZER

```python
import keras_tuner as kt
```

```python
def build_model(hp):
    model=Sequential()
    model=Sequential()
    model.add(Input(shape=(8,)))
    model.add(Dense(32,activation='relu'))
    model.add(Dense(1,activation='sigmoid'))

    optimizer=hp.Choice('optimizer',values=['adam','sgd','adadelta','rmsprop'])
    
    model.compile(optimizer=optimizer,loss='binary_crossentropy',metrics=['accuracy'])

    return model
```

```python
tuner=kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5)
```

```python
tuner.search(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model_2=tuner.get_best_models(num_models=1)[0]
```

```python
model_2.summary()
```

```python
model_2.fit(X_train,y_train,batch_size=32,epochs=100,initial_epoch=6,validation_data=(X_test,y_test))
```

## Number of nodes in a layer

```python
def build_model(hp):
    model=Sequential()

    units=hp.Int('units',min_value=8,max_value=128,step=8)

    model.add(Input(shape=(8,)))
    model.add(Dense(units=units,activation='relu'))
    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

    return model
```

```python
tuner=kt.RandomSearch(build_model,objective='val_accuracy',max_trials=16,directory='smydir')
```

```python
tuner.search(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
tuner.results_summary()
```

```python
model_3=tuner.get_best_models(num_models=1)[0]
```

```python
model_3.fit(X_train,y_train,epochs=100,initial_epoch=6,batch_size=32,validation_data=(X_test,y_test))
```

## Number of layers in NN

```python
def build_model(hp):
    model=Sequential()
    model.add(Input(shape=(8,)))
    for i in range(hp.Int('num_layers',min_value=1,max_value=10)):
        model.add(Dense(56,activation='relu'))

    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
    return model
```

```python
tuner=kt.RandomSearch(build_model,objective='val_accuracy',max_trials=10,directory='mydir',project_name='num_layers')
```

```python
tuner.search(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
tuner.results_summary()
```

```python
model_4=tuner.get_best_models(num_models=1)[0]
```

```python
model_4.fit(X_train,y_train,batch_size=32,epochs=100,initial_epoch=6,validation_data=(X_test,y_test))
```
